```python
import MDAnalysisTests.datafiles as d
import MDAnalysis as mda

user_defined_entities = [name for name in dir(d) if not name.startswith('__') and not name.endswith('__')]

for name in user_defined_entities:
    variable_value = getattr(d, name)
    try:
        u = mda.Universe(variable_value)
        if len(u.atoms) == 712:
            print(variable_value)
        
    except:
        continue
print("Done")
        
```

```python
import MDAnalysisTests.datafiles as d
import MDAnalysis as mda

user_defined_entities = [name for name in dir(d) if not name.startswith('__') and not name.endswith('__')]

for name in user_defined_entities:
    variable_value = getattr(d, name)
    try:
        u = mda.Universe(variable_value)
        sel = u.select_atoms("segid DPMC")
        if repr(sel) != "<AtomGroup with 0 atoms>":
            print(name)
        
    except:
        continue
print("Done")
        
```
